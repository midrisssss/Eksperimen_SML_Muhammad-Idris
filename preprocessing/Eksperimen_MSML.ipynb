# **1. Perkenalan Dataset**


**Dataset**: Wine Quality Dataset (Red Wine)
**Sumber**: UCI Machine Learning Repository
**Deskripsi**: Dataset berisi sifat fisikokimia sampel anggur merah dan label kualitasnya.
**Target**: 'quality_group' (0: Low/Medium, 1: High Quality)


# **2. Import Library**

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import os
```

# **3. Memuat Dataset**

```python
url = "https://raw.githubusercontent.com/mlflow/mlflow/master/tests/datasets/winequality-red.csv"
df_raw = pd.read_csv(url, sep=";")
print("5 Baris Pertama Raw Data:")
print(df_raw.head())

os.makedirs("../", exist_ok=True)
df_raw.to_csv("winequality_raw.csv", index=False)code here
```

# **4. Exploratory Data Analysis (EDA)**


```python
print("\nInformasi Dataset:")
print(df_raw.info())

print("\nJumlah Missing Values:")
print(df_raw.isnull().sum())

print("\nStatistik Deskriptif:")
print(df_raw.describe())

print("\nDistribusi Target (quality):")
print(df_raw['quality'].value_counts())
```

# **5. Data Preprocessing**

```python
df_clean = df_raw.drop_duplicates().copy()
df_clean = df_clean.dropna()

df_clean['target'] = (df_clean['quality'] >= 6).astype(int)
df_clean = df_clean.drop(columns=['quality'])

X = df_clean.drop(columns=['target'])
y = df_clean['target']

scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)

df_preprocessed = pd.concat([X_scaled, y.reset_index(drop=True)], axis=1)

print("\nHasil Preprocessing Data:")
print(df_preprocessed.head())

df_preprocessed.to_csv("winequality_preprocessing.csv", index=False)
```
